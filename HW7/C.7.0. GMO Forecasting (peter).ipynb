```python
import pandas as pd
import numpy as np
signals=pd.read_excel('../data/gmo_analysis_data.xlsx',sheet_name=1)
rf=pd.read_excel('../data/gmo_analysis_data.xlsx',sheet_name=2)
returns=pd.read_excel('../data/gmo_analysis_data.xlsx',sheet_name=3)
```

```python
excess_returns=returns.set_index('date').subtract(rf.set_index('date')['TBill 3M']/12,axis=0).reset_index()
excess_returns
```

```python
#2.1

sample1=excess_returns[excess_returns['date'].dt.year<=2011]
sample2=excess_returns[excess_returns['date'].dt.year>=2012]
sample3=excess_returns
samples={'before 2011':sample1,'after 2021':sample2,'full sample':sample3}

df = pd.DataFrame(index=['before 2011','after 2021','full sample'],columns=['mean','vol','sharpe'])


for time in samples:
    sample=samples[time]
    mean=sample['GMWAX'].mean()*12
    vol=sample['GMWAX'].std()*12**0.5
    sharpe=mean/vol

    df.loc[time]=[mean,vol,sharpe]

df
```

```python
#2.2

sample1_returns=returns[returns['date'].dt.year<=2011]
sample2_returns=returns[returns['date'].dt.year>=2012]
sample3_returns=returns
samples_returns={'before 2011':sample1_returns,'after 2021':sample2_returns,'full sample':sample3_returns}


def max_drawdown_from_returns(returns):

    nav = (1 + returns).cumprod()
    running_max = nav.cummax()
    drawdown = nav / running_max - 1
    return drawdown.min()  

df = pd.DataFrame(index=['before 2011','after 2021','full sample'],columns=['min','5th percentile','drawdown'])


for time in samples:
    sample=samples[time]
    min_val=sample['GMWAX'].min()
    perc_5=sample['GMWAX'].quantile(0.05)
    drawdown=max_drawdown_from_returns(samples_returns[time]['GMWAX'])
    df.loc[time]=[min_val,perc_5,drawdown]

df


```

```python
#2.3
import numpy as np
import statsmodels.api as sm
X = sm.add_constant(excess_returns[['SPY']])
res = sm.OLS(excess_returns['GMWAX'], X).fit()
alpha = res.params['const']
beta = res.params['SPY']
r2 = res.rsquared

samples={'before 2011':sample1,'after 2021':sample2,'full sample':sample3}

df = pd.DataFrame(index=['before 2011','after 2021','full sample'],columns=['alpha','beta','r2'])


for time in samples:
    sample=samples[time]
    X = sm.add_constant(sample[['SPY']])
    res = sm.OLS(sample['GMWAX'], X).fit()
    alpha = res.params['const']
    beta = res.params['SPY']
    r2 = res.rsquared

    df.loc[time]=[alpha,beta,r2]



df
```

## 3 Forecast Regressions

_This section utilizes data in `gmo_data.xlsx`._

1. **Lagged regression.** Consider the regression with predictors lagged one period:

$$
r^{SPY}_{t} \;=\; \alpha^{SPY,X} \;+\; \big(\beta^{SPY,X}\big)^\prime X_{t-1} \;+\; \epsilon^{SPY,X}_{t}
\tag{1}
$$

Estimate (1) and report the **$R^2$**, as well as the OLS estimates for $\alpha$ and $\beta$. Do this for:
   - $X$ as a single regressor, the **dividend–price** ratio ($DP$)
   - $X$ as a single regressor, the **earnings–price** ratio ($EP$)
   - $X$ with **three** regressors: $DP$, $EP$, and the **10‑year yield**  
   For each, report the **$R^2$**.

2. **Trading strategy from forecasts.** For each of the three regressions:
   - Build the forecasted SPY return: $\hat r^{SPY}_{t+1}$ (forecast made using $X_t$ to predict $r^{SPY}_{t+1}$).
   - Set the scale (portfolio weight) to $w_t = 100 \,\hat r^{SPY}_{t+1}$.
   - Strategy return: $r^x_{t+1} = w_t\, r^{SPY}_{t+1}$.  
   For each strategy, compute:
   - mean, volatility, Sharpe
   - max drawdown
   - market **alpha**
   - market **beta**
   - market **information ratio**

3. **Risk characteristics.**
   - For both strategies, the market, and GMO, compute monthly **VaR** at $\pi = 0.05$ (use the historical quantile).
   - The case mentions stocks under‑performed short‑term bonds from 2000–2011. Does the dynamic portfolio above under‑perform the risk‑free rate over this time?
   - Based on the regression estimates, in how many periods do we estimate a **negative risk premium**?
   - Do you believe the dynamic strategy takes on **extra risk**?

```python
#3.1

signals_copy=signals.copy()

regressors={'DP':['SPX D/P'],'EP':['SPX E/P'],'all':['SPX D/P','SPX E/P','T-Note 10YR']	}
signals_copy['SPY_lag']=returns['SPY'].shift(-1)
signals_copy=signals_copy.dropna()

all_cols= ['const', 'SPX D/P', 'SPX E/P', 'T-Note 10YR']

coef_df = pd.DataFrame(index=regressors.keys(), columns=all_cols, dtype=float)
r2_series = pd.Series(index=regressors.keys(), dtype=float)

for name, cols in regressors.items():
    X = sm.add_constant(signals_copy[cols])
    y = signals_copy['SPY_lag']
    res = sm.OLS(y, X).fit()

    
    for col in all_cols:
        if col in res.params.index:
            coef_df.loc[name, col] = res.params[col]
        else:
            coef_df.loc[name, col] = np.nan

    r2_series.loc[name] = res.rsquared


result = pd.concat([coef_df, r2_series.rename('R2')], axis=1)
result

```

```python
#3.2

all_cols= ['mean', 'vol', 'sharpe', 'drawdown','alpha','beta','ir']
df = pd.DataFrame(index=regressors.keys(), columns=all_cols, dtype=float)

df2=pd.DataFrame(index=['market','GMWAX','DP','EP','all'], columns=['var(0.05)'], dtype=float)

for name, cols in regressors.items():
    X = sm.add_constant(signals_copy[cols])
    y = signals_copy['SPY_lag']
    res = sm.OLS(y, X).fit()
    pre=res.predict(X)
    strategy_returns=pre*100*y
    mean=strategy_returns.mean()*12
    vol=strategy_returns.std()*12**0.5
    sharpe=(strategy_returns-rf[1:]['TBill 3M']/12).mean()*12/vol
    drawdown=max_drawdown_from_returns(strategy_returns)

    X=sm.add_constant(returns[1:]['SPY'].reset_index())
    y=strategy_returns

    
    res = sm.OLS(y, X).fit()

    alpha = res.params['const']
    beta = res.params['SPY']

    ir=alpha/res.resid.std()
    
    df.loc[name]=[mean,vol,sharpe,drawdown,alpha,beta,ir]
    df2.loc[name]=[strategy_returns.quantile(0.05)]

df


```

```python
#3.3

df2.loc['market']=returns['SPY'].quantile(0.05)
df2.loc['GMWAX']=returns['GMWAX'].quantile(0.05)

df2
```

## 4 Out‑of‑Sample Forecasting

_This section utilizes data in `gmo_data.xlsx`._ Focus on using **both** $DP$ and $EP$ as signals in (1). Compute **out‑of‑sample** ($OOS$) statistics:

**Procedure (rolling OOS):**
- Start at $t=60$.
- Estimate (1) using data **through** time $t$.
- Using the estimated parameters and $x_t$, compute the forecast for $t+1$:
  
  $$
  \hat r^{SPY}_{t+1} \;=\; \hat \alpha^{SPY,X}_t \;+\; \big(\hat \beta^{SPY,X}_t\big)^\prime x_t
  $$

- Forecast error: $e^{forecast}_{t+1} = r^{SPY}_{t+1} - \hat r^{SPY}_{t+1}$.
- Move to $t=61$ and iterate.

Also compute the **null** forecast and errors:

$$
\bar r^{SPY}_{t+1} = \frac{1}{t}\sum_{i=1}^t r^{SPY}_i, \qquad
e^{null}_{t+1} = r^{SPY}_{t+1} - \bar r^{SPY}_{t+1}.
$$

1. **Report the out‑of‑sample $R^2$**

$$
R^2_{OOS} \;\equiv\; 1 - \frac{\sum_{i=61}^T \big(e^{forecast}_i\big)^2}{\sum_{i=61}^T \big(e^{null}_i\big)^2}
$$

Did this forecasting strategy produce a positive $R^2_{OOS}$?

2. **Redo 3.2 with OOS forecasts.** How does the OOS strategy compare to the in‑sample version of 3.2?

3. **Redo 3.3 with OOS forecasts.** Is the point‑in‑time version of the strategy **riskier**?

```python
#4.1
error_list=[]
null_list=[]
for t in range(60,len(signals_copy)):
    

    X = sm.add_constant(signals_copy[['SPX D/P','SPX E/P']][0:t])
    y = signals_copy['SPY_lag'][0:t]
        
    res = sm.OLS(y, X).fit()
    test=sm.add_constant(signals_copy[['SPX D/P','SPX E/P']][t:t+1],has_constant='add')
    pre=res.predict(test)
    error_list.append(pre-signals_copy['SPY_lag'][t])
    null_list.append(signals_copy['SPY_lag'][0:t].mean()-signals_copy['SPY_lag'][t])
error_list=np.array(error_list)
null_list=np.array(null_list)

r2=1-np.sum(error_list**2)/np.sum(null_list**2)
r2
   
```

```python
#4.2

all_cols= ['mean', 'vol', 'sharpe', 'drawdown','alpha','beta','ir']
df = pd.DataFrame(index=regressors.keys(), columns=all_cols, dtype=float)

df2=pd.DataFrame(index=['market','GMWAX','DP','EP','all'], columns=['var(0.05)'], dtype=float)

for name, cols in regressors.items():

    strategy_returns=[]

    for t in range(60,len(signals_copy)):
        X = sm.add_constant(signals_copy[cols][0:t])
        y = signals_copy['SPY_lag'][0:t]
        
        res = sm.OLS(y, X).fit()
        test=sm.add_constant(signals_copy[cols][t:t+1],has_constant='add')
        pre=res.predict(test).iloc[0]
        
        strategy_returns.append(pre*100*signals_copy['SPY_lag'][t:t+1].iloc[0])
        
        
    
    strategy_returns=np.array(strategy_returns)
    mean=strategy_returns.mean()*12
    vol=strategy_returns.std()*12**0.5

    
    sharpe=(strategy_returns-rf[61:]['TBill 3M']/12).mean()*12/vol
    
    drawdown=max_drawdown_from_returns(pd.Series(strategy_returns))

    X=sm.add_constant(returns[61:]['SPY'].reset_index())
    y=strategy_returns

    res = sm.OLS(y, X).fit()

    alpha = res.params['const']
    beta = res.params['SPY']

    ir=alpha/res.resid.std()
    
    df.loc[name]=[mean,vol,sharpe,drawdown,alpha,beta,ir]
    df2.loc[name]=[pd.Series(strategy_returns).quantile(0.05)]

df


```

```python
#4.3

df2.loc['market']=returns['SPY'].quantile(0.05)
df2.loc['GMWAX']=returns['GMWAX'].quantile(0.05)

df2
```

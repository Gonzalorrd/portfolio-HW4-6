```python
import pandas as pd
import numpy as np
import statsmodels.api as sm


file_path = "../data/gmo_analysis_data.xlsx"

xls = pd.ExcelFile(file_path)
ret_df = pd.read_excel(xls, sheet_name="total returns")
rf_df  = pd.read_excel(xls, sheet_name="risk-free rate")

df = pd.merge(ret_df, rf_df, on="date", how="inner")
df = df.sort_values("date").set_index("date")

df["rf_m"] = df["TBill 3M"] / 12.0

df["SPY_ex"]   = df["SPY"]   - df["rf_m"]
df["GMWAX_ex"] = df["GMWAX"] - df["rf_m"]
df["GMGEX_ex"] = df["GMGEX"] - df["rf_m"]

FREQ = 12   # monthly


cutoff_2011 = pd.Timestamp("2011-12-31")
start_2012  = pd.Timestamp("2012-01-31")

samples = {
    "incept_2011": (df.index <= cutoff_2011),
    "2012_present": (df.index >= start_2012),
    "full": np.ones(len(df), dtype=bool),
}

def perf_stats(series_ex):
    mean_m = series_ex.mean()
    vol_m  = series_ex.std(ddof=1)
    mean_a = mean_m * FREQ
    vol_a  = vol_m * np.sqrt(FREQ)
    sharpe = mean_a / vol_a if vol_a != 0 else np.nan
    return mean_a, vol_a, sharpe


def tail_stats(series_total):
    min_ret = series_total.min()
    var5    = np.quantile(series_total, 0.05)
    wealth  = (1 + series_total).cumprod()
    rollmax = wealth.cummax()
    dd      = wealth / rollmax - 1
    max_dd  = dd.min()
    return min_ret, var5, max_dd


def capm_reg(y_ex, x_ex):
    X = sm.add_constant(x_ex)
    model = sm.OLS(y_ex, X).fit()
    alpha_m = model.params["const"]
    beta     = model.params[x_ex.name]
    alpha_a  = alpha_m * FREQ
    return {
        "alpha_ann": alpha_a,
        "beta": beta,
        "r2": model.rsquared,
        "t_alpha": model.tvalues["const"],
        "t_beta": model.tvalues[x_ex.name],
    }


results = {
    "GMWAX": {"perf": {}, "tail": {}, "reg": {}},
    "GMGEX": {"perf": {}, "tail": {}, "reg": {}},
}


for name, mask in samples.items():

    ex_spy = df.loc[mask, "SPY_ex"]


    ex_gmwax  = df.loc[mask, "GMWAX_ex"]
    tot_gmwax = df.loc[mask, "GMWAX"]

    results["GMWAX"]["perf"][name] = perf_stats(ex_gmwax)
    results["GMWAX"]["tail"][name] = tail_stats(tot_gmwax)
    results["GMWAX"]["reg"][name]  = capm_reg(ex_gmwax, ex_spy)


    ex_gmgex  = df.loc[mask, "GMGEX_ex"]
    tot_gmgex = df.loc[mask, "GMGEX"]

    results["GMGEX"]["perf"][name] = perf_stats(ex_gmgex)
    results["GMGEX"]["tail"][name] = tail_stats(tot_gmgex)
    results["GMGEX"]["reg"][name]  = capm_reg(ex_gmgex, ex_spy)


idx_samples = ["incept_2011", "2012_present", "full"]

def build_perf_table(fund):
    rows = []
    for s in idx_samples:
        rows.append(list(results[fund]["perf"][s]))
    return pd.DataFrame(rows, index=idx_samples,
                        columns=["Mean_ann_ex", "Vol_ann_ex", "Sharpe"])


def build_tail_table(fund):
    rows = []
    for s in idx_samples:
        rows.append(list(results[fund]["tail"][s]))
    return pd.DataFrame(rows, index=idx_samples,
                        columns=["Min_month", "VaR_5", "Max_drawdown"])


def build_reg_table(fund):
    rows = []
    for s in idx_samples:
        r = results[fund]["reg"][s]
        rows.append([r["alpha_ann"], r["beta"], r["r2"], r["t_alpha"], r["t_beta"]])
    return pd.DataFrame(rows, index=idx_samples,
                        columns=["Alpha_ann", "Beta", "R2", "t_alpha", "t_beta"])

```

1. **Lagged regression.** Consider the regression with predictors lagged one period:

$$

$$

Estimate (1) and report the **$R^2$**, as well as the OLS estimates for $\alpha$ and $\beta$. Do this for:
   - $X$ as a single regressor, the **dividend–price** ratio ($DP$)
   - $X$ as a single regressor, the **earnings–price** ratio ($EP$)

```python
y = reg_df["SPY"]     

# (a) X = DP only
model_DP = run_reg(y, reg_df[["DP_lag"]])
alpha_DP  = model_DP.params["const"]
beta_DP   = model_DP.params["DP_lag"]
r2_DP     = model_DP.rsquared

# (b) X = EP only
model_EP = run_reg(y, reg_df[["EP_lag"]])
alpha_EP  = model_EP.params["const"]
beta_EP   = model_EP.params["EP_lag"]
r2_EP     = model_EP.rsquared

# (c) X = (DP, EP, 10-year yield)
model_all = run_reg(y, reg_df[["DP_lag", "EP_lag", "Y10_lag"]])
alpha_all = model_all.params["const"]
beta_all  = model_all.params[["DP_lag", "EP_lag", "Y10_lag"]]
r2_all    = model_all.rsquared

print("X = DP:")
print(f"  alpha = {alpha_DP:.6f},  beta_DP = {beta_DP:.6f},  R^2 = {r2_DP:.4f}\n")

print("X = EP:")
print(f"  alpha = {alpha_EP:.6f},  beta_EP = {beta_EP:.6f},  R^2 = {r2_EP:.4f}\n")

print("X = (DP, EP, 10Y):")
print(f"  alpha = {alpha_all:.6f}")
print(f"  beta_DP  = {beta_all['DP_lag']:.6f}")
print(f"  beta_EP  = {beta_all['EP_lag']:.6f}")
print(f"  beta_10Y = {beta_all['Y10_lag']:.6f}")
print(f"  R^2      = {r2_all:.4f}\n")
```

2. **Trading strategy from forecasts.** For each of the three regressions:
   - Build the forecasted SPY return: $\hat r^{SPY}_{t+1}$ (forecast made using $X_t$ to predict $r^{SPY}_{t+1}$).
   - Set the scale (portfolio weight) to $w_t = 100 \,\hat r^{SPY}_{t+1}$.
   - Strategy return: $r^x_{t+1} = w_t\, r^{SPY}_{t+1}$.  
   For each strategy, compute:
   - mean, volatility, Sharpe
   - max drawdown
   - market **alpha**
   - market **beta**
   - market **information ratio**


```python

df = (ret_df
      .merge(rf_df, on="date", how="inner")
      .merge(signals, on="date", how="inner"))
df = df.sort_values("date").set_index("date")



df["rf_m"]   = df["TBill 3M"] / 12.0
df["SPY_ex"] = df["SPY"] - df["rf_m"]

df["DP_lag"]  = df["SPX D/P"].shift(1)
df["EP_lag"]  = df["SPX E/P"].shift(1)
df["Y10_lag"] = df["T-Note 10YR"].shift(1)

reg_df = df.dropna(subset=["DP_lag", "EP_lag", "Y10_lag"]).copy()
```

```python
y = reg_df["SPY"]

def run_reg(y, X):
    Xc = sm.add_constant(X)
    return sm.OLS(y, Xc).fit()

model_DP  = run_reg(y, reg_df[["DP_lag"]])
model_EP  = run_reg(y, reg_df[["EP_lag"]])
model_all = run_reg(y, reg_df[["DP_lag", "EP_lag", "Y10_lag"]])

reg_df["rhat_DP"]  = model_DP.fittedvalues
reg_df["rhat_EP"]  = model_EP.fittedvalues
reg_df["rhat_all"] = model_all.fittedvalues

for key in ["DP", "EP", "all"]:
    reg_df[f"w_{key}"]         = 100.0 * reg_df[f"rhat_{key}"]
    reg_df[f"ret_strat_{key}"] = reg_df[f"w_{key}"] * reg_df["SPY"]
```

```python
def strat_metrics(ret_strat, reg_df_local):
    rf_series = reg_df_local["rf_m"]
    mkt       = reg_df_local["SPY"]

    # annualized mean and vol
    mean_m = ret_strat.mean()
    vol_m  = ret_strat.std(ddof=1)
    mean_a = mean_m * FREQ
    vol_a  = vol_m * np.sqrt(FREQ)

    sharpe = (mean_m - rf_series.mean()) * FREQ / vol_a if vol_a != 0 else np.nan

    # max drawdown
    wealth  = (1 + ret_strat).cumprod()
    rollmax = wealth.cummax()
    dd      = wealth / rollmax - 1
    max_dd  = dd.min()

    # CAPM vs market
    y_ex = ret_strat - rf_series
    x_ex = mkt       - rf_series
    x_ex = x_ex.copy()
    x_ex.name = "SPY_ex_for_reg"

    X = sm.add_constant(x_ex)
    model = sm.OLS(y_ex, X).fit()

    alpha_m = model.params["const"]
    beta    = model.params["SPY_ex_for_reg"]
    alpha_a = alpha_m * FREQ

    resid_vol_a = model.resid.std(ddof=1) * np.sqrt(FREQ)
    info_ratio  = alpha_a / resid_vol_a if resid_vol_a != 0 else np.nan

    return dict(
        mean_ann   = mean_a,
        vol_ann    = vol_a,
        sharpe     = sharpe,
        max_dd     = max_dd,
        alpha_ann  = alpha_a,
        beta       = beta,
        info_ratio = info_ratio,
    )

metrics_DP  = strat_metrics(reg_df["ret_strat_DP"],  reg_df)
metrics_EP  = strat_metrics(reg_df["ret_strat_EP"],  reg_df)
metrics_all = strat_metrics(reg_df["ret_strat_all"], reg_df)
```

```python
for name, m in [("DP", metrics_DP), ("EP", metrics_EP), ("DP+EP+10Y", metrics_all)]:
    print(f"Strategy based on {name}:")
    print(f"  mean (ann)   = {m['mean_ann']:.4f}")
    print(f"  vol (ann)    = {m['vol_ann']:.4f}")
    print(f"  Sharpe       = {m['sharpe']:.4f}")
    print(f"  max drawdown = {m['max_dd']:.4f}")
    print(f"  alpha (ann)  = {m['alpha_ann']:.4f}")
    print(f"  beta         = {m['beta']:.4f}")
    print(f"  info ratio   = {m['info_ratio']:.4f}\n")
```

## 4 Out‑of‑Sample Forecasting

_This section utilizes data in `gmo_data.xlsx`._ Focus on using **both** $DP$ and $EP$ as signals in (1). Compute **out‑of‑sample** ($OOS$) statistics:

**Procedure (rolling OOS):**
- Start at $t=60$.
- Estimate (1) using data **through** time $t$.
- Using the estimated parameters and $x_t$, compute the forecast for $t+1$:
  
  $$
  \hat r^{SPY}_{t+1} \;=\; \hat \alpha^{SPY,X}_t \;+\; \big(\hat \beta^{SPY,X}_t\big)^\prime x_t
  $$

- Forecast error: $e^{forecast}_{t+1} = r^{SPY}_{t+1} - \hat r^{SPY}_{t+1}$.
- Move to $t=61$ and iterate.

Also compute the **null** forecast and errors:

$$
\bar r^{SPY}_{t+1} = \frac{1}{t}\sum_{i=1}^t r^{SPY}_i, \qquad
e^{null}_{t+1} = r^{SPY}_{t+1} - \bar r^{SPY}_{t+1}.
$$


```python

FREQ = 12
work = df[["SPY", "SPX D/P", "SPX E/P", "rf_m"]].dropna().copy()
work["r_fut"] = work["SPY"].shift(-1)
work = work.dropna(subset=["r_fut"]).copy()

y = work["r_fut"]
X = work[["SPX D/P", "SPX E/P"]]

forecast_list = []
null_list     = []
realized_list = []
date_list     = []

for i in range(60, len(work)):
    y_tr = y.iloc[:i]          
    X_tr = X.iloc[:i]
    X_tr_c = sm.add_constant(X_tr)

    model = sm.OLS(y_tr, X_tr_c).fit()

    x_now = X.iloc[i:i+1]
    x_now_c = sm.add_constant(x_now, has_constant="add")
    y_hat = float(model.predict(x_now_c))

    y_real = y.iloc[i]
    y_null = y_tr.mean()

    forecast_list.append(y_hat)
    null_list.append(y_null)
    realized_list.append(y_real)
    date_list.append(y.index[i])

oos_df = pd.DataFrame(
    {
        "SPY":      realized_list,
        "rhat_oos": forecast_list,
        "rbar_oos": null_list,
    },
    index=date_list,
)

oos_df["e_forecast"] = oos_df["SPY"] - oos_df["rhat_oos"]
oos_df["e_null"]     = oos_df["SPY"] - oos_df["rbar_oos"]

ss_forecast = np.sum(oos_df["e_forecast"] ** 2)
ss_null     = np.sum(oos_df["e_null"] ** 2)

R2_oos = 1.0 - ss_forecast / ss_null
print(f"OOS R^2 using DP and EP = {R2_oos:.4f}")
```

1. **Report the out‑of‑sample $R^2$**

$$
R^2_{OOS} \;\equiv\; 1 - \frac{\sum_{i=61}^T \big(e^{forecast}_i\big)^2}{\sum_{i=61}^T \big(e^{null}_i\big)^2}
$$

Did this forecasting strategy produce a positive $R^2_{OOS}$?



```python
print("OOS R2 =", R2_oos)
```

No it produces a negative r squared value

2. **Redo 3.2 with OOS forecasts.** How does the OOS strategy compare to the in‑sample version of 3.2?



```python

FREQ = 12

rf_shifted = df["rf_m"].shift(-1)
rf_oos = rf_shifted.loc[oos_df.index]

oos_df["rf_m"] = rf_oos

oos_df["w_oos"]         = 100.0 * oos_df["rhat_oos"]
oos_df["ret_strat_oos"] = oos_df["w_oos"] * oos_df["SPY"]

def strat_metrics_oos(ret_strat, frame):
    rf_series = frame["rf_m"]
    mkt       = frame["SPY"]

    mean_m = ret_strat.mean()
    vol_m  = ret_strat.std(ddof=1)
    mean_a = mean_m * FREQ
    vol_a  = vol_m * np.sqrt(FREQ)

    sharpe = (mean_m - rf_series.mean()) * FREQ / vol_a if vol_a != 0 else np.nan

    wealth  = (1 + ret_strat).cumprod()
    rollmax = wealth.cummax()
    dd      = wealth / rollmax - 1.0
    max_dd  = dd.min()

    y_ex = ret_strat - rf_series
    x_ex = mkt       - rf_series
    x_ex = x_ex.copy()
    x_ex.name = "SPY_ex_for_reg"

    Xc = sm.add_constant(x_ex)
    model = sm.OLS(y_ex, Xc).fit()

    alpha_m = model.params["const"]
    beta    = model.params["SPY_ex_for_reg"]
    alpha_a = alpha_m * FREQ

    resid_vol_a = model.resid.std(ddof=1) * np.sqrt(FREQ)
    info_ratio  = alpha_a / resid_vol_a if resid_vol_a != 0 else np.nan

    return dict(
        mean_ann   = mean_a,
        vol_ann    = vol_a,
        sharpe     = sharpe,
        max_dd     = max_dd,
        alpha_ann  = alpha_a,
        beta       = beta,
        info_ratio = info_ratio,
    )

metrics_oos = strat_metrics_oos(oos_df["ret_strat_oos"], oos_df)

print("OOS strategy based on DP and EP:")
for k, v in metrics_oos.items():
    print(f"{k:10s}: {v:.4f}")
```

The OOS strategy performs much worse than the in-sample versions. Its Sharpe falls by about half, alpha turns negative, the information ratio becomes negative, and the max drawdown is slightly worse. In contrast, all in-sample strategies showed higher returns, higher Sharpe ratios, and positive alpha. This shows the forecast model does not generalize well out-of-sample.

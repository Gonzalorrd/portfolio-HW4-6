```python
#Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm
```

```python
#2.1
path = "../data/gmo_analysis_data.xlsx"
df_returns = pd.read_excel(path, sheet_name="total returns")
df_rf = pd.read_excel(path, sheet_name="risk-free rate")

df_returns['date']=pd.to_datetime(df_returns["date"])
df_rf["date"]=pd.to_datetime(df_rf["date"])

df_merged=pd.merge(df_returns,df_rf, on='date')
df_merged['excess_GMWAX']=df_merged['GMWAX']-df_merged['TBill 3M']

df_merged['date'] = pd.to_datetime(df_merged['date'])

def stats(sub, string='excess_GMWAX'):
    x = sub[string]
    mean = x.mean()*12
    vol  = x.std(ddof=0)*np.sqrt(12)
    sharpe = np.nan if vol==0 else mean/vol
    return pd.Series({
        'mean': mean, 'vol': vol, 'Sharpe': sharpe,
    })


out = pd.DataFrame({
    'inception - 2011': stats(df_merged[df_merged['date'] <= '2011-12-31']),
    '2012 - present': stats(df_merged[(df_merged['date'] >= '2012-01-01') & (df_merged['date'] <= '2025-12-31')]),
    'inception - present': stats(df_merged),
}).T
out
```

the mean return is still negative in all periods, but it’s less bad after 2012 and volatility is a bit lower. As a result, the Sharpe ratio has improved. Overall, performance has gotten somewhat better, but not enough to change my conclusion that the strategy remains unattractive on a risk-adjusted basis.

```python
#2.2

def calc_max_drawdown(returns: pd.Series, initial_wealth: float = 1000.0) -> float:
    wealth_index = initial_wealth * (1 + returns).cumprod()
    previous_peaks = wealth_index.cummax()
    drawdowns = (wealth_index - previous_peaks) / previous_peaks
    return drawdowns.min()

# Define subsamples
subsamples = {
    "inception - 2011":    df_merged.loc[:'2011-12-31'],
    "2012 - present":      df_merged.loc['2012-01-01':],
    "inception - present": df_merged
}

rows = []
for name, sub in subsamples.items():
    minimum = sub['excess_GMWAX'].min()
    VaR_5   = sub['excess_GMWAX'].quantile(q=0.05)
    drawdown = calc_max_drawdown(sub['GMWAX'])

    rows.append({
        "sample": name,
        "min_return": minimum,
        "VaR_5": VaR_5,
        "max_drawdown": drawdown
    })

out = pd.DataFrame(rows).set_index("sample")
print(out)
```

a)GMVAX shows meaningful tail risk. The worst monthly loss is almost -20%, the 5% VaR is about -8%, and the max drawdown is close to -30%, so big losses are very much on the table.

b) These tail-risk measures are of similar magnitude across the two subsamples, so tail risk does not appear to vary much over time.

```python
#2.3
df_merged['excess_SPY'] = df_merged['SPY'] - df_merged['TBill 3M']

def regress_two_series(y: pd.Series, x: pd.Series) -> pd.Series:
    df = pd.concat([y, x], axis=1, keys=['y', 'x']).dropna()
    X = sm.add_constant(df['x'])
    model = sm.OLS(df['y'], X).fit()
    return pd.Series({
        'alpha': model.params['const'],
        'beta':  model.params['x'],
        'r2':    model.rsquared
    })

# Define subsamples
subsamples = {
    "inception - 2011":    df_merged.loc[:'2011-12-31'],
    "2012 - present":      df_merged.loc['2012-01-01':],
    "inception - present": df_merged
}

rows = []
for name, sub in subsamples.items():
    res = regress_two_series(sub['excess_GMWAX'], sub['excess_SPY'])
    res['sample'] = name
    rows.append(res)

reg_out = pd.DataFrame(rows).set_index('sample')
print(reg_out)
```

-GMWAX still looks like a low-beta strategy: the beta is around 0.62, and it stays well below 1 in both subsamples, so its market exposure hasn’t really changed since the case.

-On alpha, the regression gives a slightly negative alpha  and that pattern is similar across the subsamples, so I don’t see evidence that GMWAX has delivered positive alpha either in the original case period or more recently.

```python
#2.4
df_merged['excess_GMGEX'] = df_merged['GMGEX'] - df_merged['TBill 3M']

out = pd.DataFrame({
    'inception - 2011': stats(df_merged[df_merged['date']<= '2011-12-31'], string='excess_GMGEX'),
    '2012 - present':   stats(df_merged[(df_merged['date'] >= '2012-01-01') & (df_merged['date'] <= '2025-12-31')], string='excess_GMGEX'),
    'inception - present': stats(df_merged, string='excess_GMGEX'),
}).T


subsamples = {
    "inception - 2011":    df_merged.loc[:'2011-12-31'],
    "2012 - present":      df_merged.loc['2012-01-01':],
    "inception - present": df_merged
}


tail_rows = []
reg_rows = []

for name, sub in subsamples.items():
    minimum_GEX = sub['excess_GMGEX'].min()
    VaR_5_GEX   = sub['excess_GMGEX'].quantile(q=0.05)
    drawdown_GEX = calc_max_drawdown(sub['GMGEX'])

    tail_rows.append({
        "sample": name,
        "min_return": minimum_GEX,
        "VaR_5": VaR_5_GEX,
        "max_drawdown": drawdown_GEX
    })

    reg = regress_two_series(sub['excess_GMGEX'], sub['excess_SPY'])
    reg['sample'] = name
    reg_rows.append(reg)

tail_GEX = pd.DataFrame(tail_rows).set_index("sample")
reg_GEX  = pd.DataFrame(reg_rows).set_index("sample")

print("summary statistics:\n", out)
print("\nTail-risk metrics:\n", tail_GEX)
print("\nRegression results (GMGEX vs SPY):\n", reg_GEX)
```

```python
#3.1
df_signals = pd.read_excel(path, sheet_name="signals")

reg_2 = regress_two_series(df_merged['SPY'], df_signals['SPX D/P'].shift(1))
reg_3 = regress_two_series(df_merged['SPY'], df_signals['SPX E/P'].shift(1))

def regress_on_df(y: pd.Series, x: pd.DataFrame) -> pd.Series:
    df = pd.concat([y.rename('y'), x], axis=1).dropna()

    y_num = pd.to_numeric(df['y'], errors='raise')
    X_num = df.drop(columns=['y', 'date']).apply(pd.to_numeric, errors='raise')

    X = sm.add_constant(X_num)
    model = sm.OLS(y_num, X).fit()
    out = pd.Series({
        'alpha': model.params['const'],
        'r2':    model.rsquared
    })
    betas = model.params.drop('const').rename(lambda c: f'beta_{c}')
    return pd.concat([out, betas])

reg_4=regress_on_df(df_merged['SPY'], df_signals.shift(1))

results = pd.DataFrame(
    [reg_2, reg_3, reg_4],
    index=['DP', 'EP', 'ALL']
)

print(results)
```

```python
#3.2 
# ========= 1. Build lagged signals =========
# only lag the signal columns, not the date
signals_lag = df_signals[['SPX D/P', 'SPX E/P', 'T-Note 10YR']].shift(1)

# ========= 2. Build dataframes for each regression (aligned, no NaNs) =========
# DP: SPY and lagged D/P
df_DP = pd.concat([df_merged['SPY'], signals_lag['SPX D/P']], axis=1).dropna()
df_DP.columns = ['SPY', 'DP']

# EP: SPY and lagged E/P
df_EP = pd.concat([df_merged['SPY'], signals_lag['SPX E/P']], axis=1).dropna()
df_EP.columns = ['SPY', 'EP']

# ALL: SPY and all three lagged signals
df_ALL = pd.concat(
    [df_merged['SPY'],
     signals_lag[['SPX D/P', 'SPX E/P', 'T-Note 10YR']]],
    axis=1
).dropna()
df_ALL.columns = ['SPY', 'DP', 'EP', 'T10']

# ========= 3. Take coefficients from your 'results' table =========
alpha_DP = results.loc['DP', 'alpha']
beta_DP  = results.loc['DP', 'beta']

alpha_EP = results.loc['EP', 'alpha']
beta_EP  = results.loc['EP', 'beta']

alpha_ALL    = results.loc['ALL', 'alpha']
beta_DP_ALL  = results.loc['ALL', 'beta_SPX D/P']
beta_EP_ALL  = results.loc['ALL', 'beta_SPX E/P']
beta_T10_ALL = results.loc['ALL', 'beta_T-Note 10YR']

# ========= 4. Forecasts and strategy returns =========
# DP strategy
rhat_DP = alpha_DP + beta_DP * df_DP['DP']
w_DP    = 100 * rhat_DP
rstr_DP = w_DP * df_DP['SPY']

# EP strategy
rhat_EP = alpha_EP + beta_EP * df_EP['EP']
w_EP    = 100 * rhat_EP
rstr_EP = w_EP * df_EP['SPY']

# ALL strategy
rhat_ALL = (alpha_ALL
            + beta_DP_ALL  * df_ALL['DP']
            + beta_EP_ALL  * df_ALL['EP']
            + beta_T10_ALL * df_ALL['T10'])
w_ALL    = 100 * rhat_ALL
rstr_ALL = w_ALL * df_ALL['SPY']
import numpy as np
import pandas as pd

# ===== helper: compute all stats for one strategy =====
def strat_stats(r_strat: pd.Series,
                r_mkt: pd.Series,
                periods_per_year: int = 12) -> pd.Series:
    # align and drop NaNs
    df = pd.concat(
        [r_strat.rename('r'), r_mkt.rename('m')],
        axis=1
    ).dropna()

    if len(df) == 0:
        return pd.Series({
            'mean': np.nan, 'vol': np.nan, 'Sharpe': np.nan,
            'max_drawdown': np.nan,
            'alpha_mkt': np.nan, 'beta_mkt': np.nan, 'IR_mkt': np.nan
        })

    r = df['r'].to_numpy()
    m = df['m'].to_numpy()

    # ----- basic performance -----
    mean_m = r.mean() * periods_per_year
    vol_m  = r.std(ddof=1) * np.sqrt(periods_per_year)
    sharpe = np.nan if vol_m == 0 else mean_m / vol_m

    # ----- max drawdown -----
    wealth = np.cumprod(1.0 + r)
    peak   = np.maximum.accumulate(wealth)
    drawdowns = wealth / peak - 1.0
    max_dd = drawdowns.min()

    # ----- market beta & alpha -----
    cov_rm = np.cov(r, m, ddof=1)[0, 1]
    var_m  = np.var(m, ddof=1)
    if var_m == 0:
        beta = np.nan
    else:
        beta = cov_rm / var_m

    mean_r   = r.mean()
    mean_mkt = m.mean()
    if np.isnan(beta):
        alpha_month = np.nan
    else:
        alpha_month = mean_r - beta * mean_mkt

    alpha_ann = alpha_month * periods_per_year if not np.isnan(alpha_month) else np.nan

    # ----- information ratio (alpha / residual vol) -----
    if np.isnan(beta) or np.isnan(alpha_month):
        IR = np.nan
    else:
        fitted = alpha_month + beta * m
        resid  = r - fitted
        sigma_res = resid.std(ddof=1)
        IR = (alpha_month / sigma_res) * np.sqrt(periods_per_year) if sigma_res != 0 else np.nan

    return pd.Series({
        'mean': mean_m,
        'vol': vol_m,
        'Sharpe': sharpe,
        'max_drawdown': max_dd,
        'alpha_mkt': alpha_ann,
        'beta_mkt': beta,
        'IR_mkt': IR
    })

# ====== apply to the three strategies and print table ======
stats_DP  = strat_stats(rstr_DP,  df_DP['SPY'])
stats_EP  = strat_stats(rstr_EP,  df_EP['SPY'])
stats_ALL = strat_stats(rstr_ALL, df_ALL['SPY'])

summary = pd.DataFrame(
    [stats_DP, stats_EP, stats_ALL],
    index=['DP', 'EP', 'ALL']
)

print(summary.round(4))
```

```python
#3.3
def hist_var(ret: pd.Series, alpha: float = 0.05, as_loss: bool = True):
    """
    Historical VaR at level alpha from a return series.
    If as_loss=True, returns a positive loss number.
    """
    ret = ret.dropna()
    q = ret.quantile(alpha)          # 5% return quantile (usually negative)
    return -q if as_loss else q

# ---- collect series you care about ----
# adjust names to whatever you actually have
series_dict = {
    "DP":  rstr_DP,                  # DP strategy returns
    "EP":  rstr_EP,                  # EP strategy returns (or ALL, etc.)
    "ALL": rstr_ALL,                 # if you want all three
    "Market": df_merged["SPY"],      # market (SPY excess return)
    "GMO":   df_merged["excess_GMGEX"],  # or 'GMGEX' if you use raw fund returns
}

# ---- compute VaR(0.05) for each ----
var_results = {
    name: hist_var(ser, alpha=0.05, as_loss=True)
    for name, ser in series_dict.items()
}

var_table = pd.Series(var_results)
print(var_table.round(4))
```

## 4 Out‑of‑Sample Forecasting

_This section utilizes data in `gmo_data.xlsx`._ Focus on using **both** $DP$ and $EP$ as signals in (1). Compute **out‑of‑sample** ($OOS$) statistics:

**Procedure (rolling OOS):**
- Start at $t=60$.
- Estimate (1) using data **through** time $t$.
- Using the estimated parameters and $x_t$, compute the forecast for $t+1$:
  
  $$
  \hat r^{SPY}_{t+1} \;=\; \hat \alpha^{SPY,X}_t \;+\; \big(\hat \beta^{SPY,X}_t\big)^\prime x_t
  $$

- Forecast error: $e^{forecast}_{t+1} = r^{SPY}_{t+1} - \hat r^{SPY}_{t+1}$.
- Move to $t=61$ and iterate.

Also compute the **null** forecast and errors:

$$
\bar r^{SPY}_{t+1} = \frac{1}{t}\sum_{i=1}^t r^{SPY}_i, \qquad
e^{null}_{t+1} = r^{SPY}_{t+1} - \bar r^{SPY}_{t+1}.
$$

1. **Report the out‑of‑sample $R^2$**

$$
R^2_{OOS} \;\equiv\; 1 - \frac{\sum_{i=61}^T \big(e^{forecast}_i\big)^2}{\sum_{i=61}^T \big(e^{null}_i\big)^2}
$$

Did this forecasting strategy produce a positive $R^2_{OOS}$?

2. **Redo 3.2 with OOS forecasts.** How does the OOS strategy compare to the in‑sample version of 3.2?

3. **Redo 3.3 with OOS forecasts.** Is the point‑in‑time version of the strategy **riskier**?

```python
#4.1
df = pd.concat(
    [df_merged['SPY'], df_signals[['SPX D/P', 'SPX E/P']]],
    axis=1
).dropna()
df.columns = ['SPY', 'DP', 'EP']

df['SPY_lead'] = df['SPY'].shift(-1)      # r_{t+1}^SPY
df = df.dropna().reset_index(drop=True)

y = df['SPY_lead'].to_numpy()             # target
X = sm.add_constant(df[['DP', 'EP']].to_numpy())  # predictors x_t
r = df['SPY'].to_numpy()                  # r_t^SPY for null forecast

start = 60
T = len(df)
e_f, e_n = [], []

for t in range(start, T):
    # expanding window: use 0..t-1 to estimate
    model = sm.OLS(y[:t], X[:t, :]).fit()

    # forecast r_{t+1} with x_t
    rhat = model.predict(X[t:t+1, :])[0]
    y_true = y[t]

    e_f.append(y_true - rhat)        # forecast error

    # null forecast: mean of r_1,...,r_t
    r_bar = r[:t].mean()
    e_n.append(y_true - r_bar)       # null error

e_f = np.array(e_f)
e_n = np.array(e_n)

R2_OOS = 1 - (e_f**2).sum() / (e_n**2).sum()

print(f"OOS R-squared: {R2_OOS:.4f}")
```

```python
#4.2

# rebuild data
df_oos = pd.concat(
    [df_merged['SPY'], df_signals[['SPX D/P', 'SPX E/P']]],
    axis=1
).dropna()
df_oos.columns = ['SPY', 'DP', 'EP']

df_oos['SPY_lead'] = df_oos['SPY'].shift(-1)   # r_{t+1}^SPY
df_oos = df_oos.dropna()

y = df_oos['SPY_lead']                # target r_{t+1}^SPY
X = sm.add_constant(df_oos[['DP', 'EP']])
r = df_oos['SPY']                     # r_t^SPY (for null / info only)

start = 60
T = len(df_oos)

# rolling oos
rhat_list = []

for t in range(start, T):
    # estimate on observations
    model = sm.OLS(y.iloc[:t], X.iloc[:t, :]).fit()
    # forecast
    rhat_t1 = float(model.predict(X.iloc[t:t+1, :]))
    rhat_list.append(rhat_t1)

idx_oos = y.index[start:T]  

rhat_oos = pd.Series(rhat_list, index=idx_oos, name='rhat_oos')
spy_oos  = y.iloc[start:T]
spy_oos.index = idx_oos

# OOS trading strategy
w_oos    = 100 * rhat_oos
rstr_oos = w_oos * spy_oos
rstr_oos.name = 'rstr_oos'


stats_oos = strat_stats(rstr_oos, spy_oos)

print("4.2 – OOS strategy performance (DP & EP, expanding window):")
print(stats_oos.to_frame().T.round(4))

```
